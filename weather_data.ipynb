```python
#The first line pd.DatetimeIndex(data["Date"]) creates a DatetimeIndex object from the "Date" column in a DataFrame called data. This DatetimeIndex allows for easy manipulation and analysis of datetime data.
# The second line pd.DatetimeIndex(data["Date"]).year extracts the year component from each datetime value in the "Date" column and returns it as an array or Series containing only the years.

#So, in summary, the code creates a DatetimeIndex from the "Date" column and then extracts the year component from each datetime value in that column.
```

```python
data["Year"] = pd.DatetimeIndex(data["Date"]).year
data["Month"] = pd.DatetimeIndex(data["Date"]).month
data["Dayofmonth"] = pd.DatetimeIndex(data["Date"]).day
data.head()
```

```python
###Creating new columns
#The temperature given is in Celcius, convert it in Fahrenheit and store it in␣ a new column for it.
data["Maxtemp_F"] = data["MaxTemp"] * 9/5 +32
data.head()
```

```python
#This code calculates the Wind Chill Index (WCI) for each row in the DataFrame data using the WindGustSpeed and MinTemp columns.

from math import sqrt #This line imports the square root function sqrt from the Python math module.
def wci(x):#This defines a function named wci that takes a single argument x, which represents a row of the DataFrame
#the below lines extract the values of the 'WindGustSpeed' and 'MinTemp' columns from the current row x and assign them to the variables velocity and minTemp, respectively.
    
    velocity = x['WindGustSpeed']
    minTemp = x['MinTemp']
    return ((10 * sqrt(velocity) - velocity + 10.5)*(33-minTemp))
# calculates the wind chill index , which is the standara wind chill index formula     
data['WCI'] = data.apply(wci,axis=1) #This line applies the wci function row-wise to the DataFrame data, using axis=1 to specify that the function should be applied along the rows. It calculates the WCI for each row and assigns the result to a new column 'WCI' in the DataFrame.

data.head()
```
